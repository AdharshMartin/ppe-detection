```python
# Requires: pip install imageio-ffmpeg   (bundles a real ffmpeg binary — no
# system ffmpeg install needed. First run downloads it automatically.)
import os
import cv2
import json
import requests
import imageio_ffmpeg
from ultralytics import YOLO

# --- Setup Paths ---
model_path = "best.pt"
video_path = "input_video.mp4"
json_output_path = "dashboard_data.json"
web_video_output = "annotated_output.mp4"

# --- 1. Download Model Weights if missing ---
if not os.path.exists(model_path):
    print("Downloading model weights...")
    url = "https://huggingface.co/Hansung-Cho/yolov8-ppe-detection/resolve/main/best.pt"
    r = requests.get(url)
    with open(model_path, "wb") as f:
        f.write(r.content)

# --- 2. Load Model ---
model = YOLO(model_path)

# --- 3. Open Input Video ---
if not os.path.exists(video_path):
    raise FileNotFoundError(
        f"Input video not found at '{video_path}'. "
        f"Check the path is correct and the file was uploaded/placed there."
    )

cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    raise IOError(
        f"OpenCV could not open '{video_path}'. This usually means either the "
        f"file is corrupt/incomplete, or your OpenCV build is missing the codec "
        f"needed to decode it (common for H.265/HEVC .mp4 files on some systems). "
        f"Try: pip install opencv-python (not opencv-python-headless without "
        f"codec support), or re-encode the source video to H.264 with ffmpeg: "
        f"ffmpeg -i {video_path} -c:v libx264 -c:a aac fixed_input.mp4"
    )

width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if width == 0 or height == 0:
    cap.release()
    raise IOError(
        f"Video opened but reports 0 width/height ({width}x{height}). "
        f"The file is likely corrupt or an unsupported container/codec."
    )

if not fps or fps <= 0:
    print(f"Warning: could not read a valid FPS from source (got {fps}); defaulting to 25.0")
    fps = 25.0

print(f"Input video: {width}x{height} @ {fps:.2f}fps, ~{total_frames} frames")

# --- 4. Setup Output Writer (real H.264 MP4 — this is the part that ---
# --- actually determines whether the file plays in a browser/Streamlit) ---
# cv2.VideoWriter's 'mp4v' fourcc writes MPEG-4 Part 2 video inside an MP4
# container. Some desktop players (VLC) decode that fine, but browsers and
# Streamlit's <video> element require H.264, so the file "exists" but won't
# play. imageio-ffmpeg bundles a real ffmpeg binary (downloaded on first use,
# no system install needed) and lets us encode proper libx264/yuv420p output
# with faststart, which is what browsers expect.
out = imageio_ffmpeg.write_frames(
    web_video_output,
    size=(width, height),
    fps=fps,
    codec='libx264',
    pix_fmt_in='bgr24',       # matches OpenCV's native frame format, no color conversion needed
    pix_fmt_out='yuv420p',    # required for broad browser/player compatibility
    output_params=['-movflags', '+faststart']
)
out.send(None)  # prime the generator

# --- 5. Annotation Style Config ---
BOX_COLOR = (0, 0, 255)       # bright red (BGR)
TEXT_COLOR = (255, 255, 255)  # white
FONT = cv2.FONT_HERSHEY_SIMPLEX
FONT_SCALE = 0.5
FONT_THICKNESS = 1
BOX_THICKNESS = 2
LABEL_PADDING = 4

# --- Box Smoothing Config ---
# Raw per-frame detections wobble slightly even for a stationary object,
# which reads as "jitter". Fix: track each object across frames (so we know
# box A in frame N is the same object as box A in frame N+1), then smooth
# its coordinates with an exponential moving average instead of drawing the
# raw detection every frame.
SMOOTHING_ALPHA = 0.4   # lower = smoother/more lag, higher = snappier/less lag
MAX_MISSED_FRAMES = 10  # drop a track's smoothed state if unseen this long
smoothed_boxes = {}     # track_id -> {"box": [x1,y1,x2,y2], "missed": int}

frame_metrics = []
frame_idx = 0

print("Processing frames...")
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # persist=True keeps the tracker's internal state (and track IDs) alive
    # across calls, so the same physical object gets the same ID frame to
    # frame instead of being re-detected from scratch each time.
    results = model.track(source=frame, conf=0.25, persist=True, verbose=False)
    r = results[0]

    counts = {
        "Hardhat": 0,
        "Safety Vest": 0,
        "NO-Mask": 0,
        "Person": 0
    }

    seen_ids = set()

    for box in r.boxes:
        cls_id = int(box.cls[0])
        cls_name = model.names[cls_id]
        conf = float(box.conf[0])

        if cls_name in counts:
            counts[cls_name] += 1

        raw_box = list(map(float, box.xyxy[0]))
        track_id = int(box.id[0]) if box.id is not None else None

        if track_id is not None:
            seen_ids.add(track_id)
            if track_id in smoothed_boxes:
                prev = smoothed_boxes[track_id]["box"]
                # EMA: nudge the smoothed box toward the new raw detection
                # instead of jumping straight to it.
                smooth = [
                    SMOOTHING_ALPHA * raw_box[i] + (1 - SMOOTHING_ALPHA) * prev[i]
                    for i in range(4)
                ]
            else:
                smooth = raw_box  # first time we've seen this track, no history yet
            smoothed_boxes[track_id] = {"box": smooth, "missed": 0}
            x1, y1, x2, y2 = map(int, smooth)
        else:
            # No stable ID (tracker didn't associate it) — fall back to the
            # raw detection rather than dropping the box entirely.
            x1, y1, x2, y2 = map(int, raw_box)

        # --- Draw bounding box ---
        cv2.rectangle(frame, (x1, y1), (x2, y2), BOX_COLOR, BOX_THICKNESS)

        # --- Build label text ---
        label = f"{cls_name} {conf:.2f}"
        (text_w, text_h), baseline = cv2.getTextSize(
            label, FONT, FONT_SCALE, FONT_THICKNESS
        )

        # --- Compute label box position (above top-left corner) ---
        label_top = max(y1 - text_h - baseline - LABEL_PADDING * 2, 0)
        label_bottom = label_top + text_h + baseline + LABEL_PADDING * 2
        label_left = x1
        label_right = x1 + text_w + LABEL_PADDING * 2

        # If the label would go above the frame, draw it just below the top edge instead
        if y1 - text_h - baseline - LABEL_PADDING * 2 < 0:
            label_top = y1
            label_bottom = y1 + text_h + baseline + LABEL_PADDING * 2

        # --- Draw solid red label background ---
        cv2.rectangle(
            frame,
            (label_left, label_top),
            (label_right, label_bottom),
            BOX_COLOR,
            thickness=-1  # filled
        )

        # --- Draw white text inside label box ---
        text_x = label_left + LABEL_PADDING
        text_y = label_bottom - LABEL_PADDING - baseline // 2
        cv2.putText(
            frame,
            label,
            (text_x, text_y),
            FONT,
            FONT_SCALE,
            TEXT_COLOR,
            FONT_THICKNESS,
            cv2.LINE_AA
        )

    # --- Age out tracks we didn't see this frame; drop them after they've ---
    # --- been missing too long so memory doesn't grow unbounded ---
    for tid in list(smoothed_boxes.keys()):
        if tid not in seen_ids:
            smoothed_boxes[tid]["missed"] += 1
            if smoothed_boxes[tid]["missed"] > MAX_MISSED_FRAMES:
                del smoothed_boxes[tid]

    # --- Write annotated frame to output video ---
    out.send(frame.tobytes())

    # --- Store per-frame metrics ---
    frame_metrics.append({
        "frame": frame_idx,
        "hardhats": counts["Hardhat"],
        "vests": counts["Safety Vest"],
        "no_masks": counts["NO-Mask"],
        "persons": counts["Person"]
    })

    frame_idx += 1
    if frame_idx % 50 == 0:
        print(f"  Processed {frame_idx} frames...")

cap.release()
out.close()

if frame_idx == 0:
    raise RuntimeError(
        "No frames were read from the input video — the file opened but "
        "produced zero readable frames. It's likely corrupt or truncated."
    )

if not os.path.exists(web_video_output) or os.path.getsize(web_video_output) == 0:
    raise RuntimeError(
        f"Output video '{web_video_output}' was not created or is empty, "
        f"despite processing {frame_idx} frames. This points to an ffmpeg/"
        f"encoding failure — check the console output above for an ffmpeg error."
    )

print(f"Success! Annotated video written to: {web_video_output} "
      f"({os.path.getsize(web_video_output) / 1_000_000:.2f} MB, {frame_idx} frames)")

# --- 6. Export JSON for Dashboard ---
with open(json_output_path, "w") as f:
    json.dump(frame_metrics, f, indent=4)

print(f"Metrics exported to: {json_output_path}")
print("Processing complete!")
```
